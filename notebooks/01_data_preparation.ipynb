## Spalten Konvertieren

```python
def convert_to_int(list):
    for i in list:
        df[i] = pd.to_numeric(df[i], errors='coerce', downcast='integer').fillna(0).astype('int64')
    
    return df.info()
```

```python
list_to_convert_to_int = ['in_shazam_charts', 'streams','in_deezer_playlists' ]

convert_to_int(list_to_convert_to_int)
```
