```python
import pandas as pd
import musicbrainzngs as mb
```

### read the cleaned data in

```python
df = pd.read_csv('spotify_cleaned.csv')
```

```python
df.drop(['Unnamed: 0'], axis=1, inplace=True)
```

### Artists in einer Liste speichern

```python
artist_list = df['artist(s)_name'].tolist()
artist_list
```

### Funktion erstellen um das Herkunftsland des Künstlers mithilfe von musicbrainz zu bekommen

```python
def find_artist_country(artist_name):

    mb.set_useragent("Artist_Country", "1.0", contact="l-brahner@t-online.de")
    try:
        result = mb.search_artists(artist=artist_name, limit=1)
        if result['artist-list']:
            artist = result['artist-list'][0]
            country = artist['area']['name']

            return country
        else:
            return "Artist not found"
    except Exception as e:
        return f"Error: {e}"
```

### Funktion anwenden

```python
artist_country_list = []

for artist in artist_list:
    artist_country_list.append(find_artist_country(artist)) 
```

```python
df['artist_country'] = artist_country_list
```

```python
df
```

```python
df[df['artist_country'] == 'Artist not found']
```

### Spalten droppen mit artist not found

```python

df = df[df['artist_country'] != 'Artist not found']
```

### spalten droppen mit Error

```python
df = df[~df['artist_country'].str.contains('Error')]
```

```python
df.to_csv('spotify_angereichert_cleaned.csv')
```
